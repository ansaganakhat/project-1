```python
from pathlib import Path
import os
import sys

CURRENT = Path.cwd().resolve()
PROJECT_ROOT = next(
    (path for path in (CURRENT, *CURRENT.parents) if (path / "src").is_dir()),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Project root табылмады. Notebook-ты project папкасының ішінен ашыңыз."
    )
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)
print("PROJECT_ROOT:", PROJECT_ROOT)
print("Python:", sys.executable)
```

```python
from IPython.display import display

from src.config import YOLOV5_SPEC, YOLOV10_SPEC, print_config_status
from src.runtime_dataset import prepare_all_runtime_yamls
from src.yolo_dataset_utils import dataset_diagnostics

print_config_status(prepare_yaml=False)
runtime_yamls = prepare_all_runtime_yamls(
    [YOLOV5_SPEC, YOLOV10_SPEC],
    force=True,
)
print("Runtime YAML files:", runtime_yamls)
```

```python
for model_name, yaml_path in runtime_yamls.items():
    print(f"\n{model_name.upper()} DATASET")
    diagnostics = dataset_diagnostics(yaml_path)
    display(diagnostics)
    active = diagnostics[diagnostics["split"].isin(["train", "val"])]
    if active["images"].sum() == 0:
        raise ValueError(f"{model_name}: images табылмады")
    if active["objects"].sum() == 0:
        raise ValueError(
            f"{model_name}: labels/objects табылмады. "
            "missing_label_examples бағанын қараңыз."
        )
print("\nБарлық dataset жолдары дұрыс.")
```
