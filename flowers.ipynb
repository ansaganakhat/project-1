## Тапсырма 2. Датасетті жүктеу

`tfds.load` арқылы `tf_flowers` датасетін жүкте.
`with_info=True`, `as_supervised=True` параметрлерін қолдан.
Датасет тек `train` бөлігінен тұрады, сондықтан оны 80% оқу / 20% тест етіп бөл.

```python
# TODO: tfds.load('tf_flowers', split=[...], with_info=True, as_supervised=True)
# Кеңес: split=['train[:80%]', 'train[80%:]'] деп екі бөлік ал
(train_ds, test_ds), info = tfds.load(
    'tf_flowers',
    split=['train[:80%]', 'train[80%:]'],
    with_info=True,
    as_supervised=True
)

print(info)
```

## Тапсырма 3. Деректерді зерттеу

Класс санын және класс атауларын экранға шығар.
Кеңес: `info.features['label'].num_classes` және `.names`.

```python
# TODO: класс санын num_classes айнымалысына сақта
num_classes = info.features['label'].num_classes
class_names = info.features['label'].names

print("Класс саны:", num_classes)
print("Атаулары:", class_names)
```

## Тапсырма 4. Суреттерді көру

Оқу жиынынан 9 суретті 3x3 торда көрсет, әрқайсысының үстіне класс атауын жаз.
Кеңес: `for img, label in train_ds.take(9):`.

```python
plt.figure(figsize=(8, 8))

for i, (img, label) in enumerate(train_ds.take(9)):
    plt.subplot(3, 3, i + 1)
    plt.imshow(img)
    plt.title(class_names[int(label)])
    plt.axis('off')

plt.show()
```

## Тапсырма 5. Алдын ала өңдеу функциясы

Суреттерді бірдей өлшемге келтір (мысалы 128x128) және пиксельдерді 0..1 аралығына нормалда.
Функция (image, label) қабылдап, өңделген (image, label) қайтарсын.

```python
IMG_SIZE = 128

def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = image / 255.0
    return image, label
```

## Тапсырма 6. Конвейер дайындау (map, batch, prefetch)

`preprocess` функциясын екі жиынға да `.map()` арқылы қолдан.
Содан соң `.batch(32)` және `.prefetch(tf.data.AUTOTUNE)` қос.

```python
BATCH = 32
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.map(preprocess).batch(BATCH).prefetch(AUTOTUNE)
test_ds = test_ds.map(preprocess).batch(BATCH).prefetch(AUTOTUNE)
```

## Тапсырма 7. CNN моделін құру

`tf.keras.Sequential` арқылы конволюциялық желі жаса.
Кеңес: бірнеше `Conv2D` + `MaxPooling2D`, соңында `Flatten`, `Dense`.
Соңғы қабат — `num_classes` нейрон, `softmax` активациясы.

```python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input((IMG_SIZE, IMG_SIZE, 3)),

    layers.Conv2D(32, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Flatten(),

    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

model.summary()
```

## Тапсырма 8. Модельді компиляциялау

Оптимизатор `adam`, шығын `sparse_categorical_crossentropy`, метрика `accuracy`.
(Белгілер бүтін сан болғандықтан sparse нұсқасы керек.)

```python
# TODO: model.compile(optimizer=..., loss=..., metrics=[...])
# Тапсырма 8. Модельді компиляциялау
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

## Тапсырма 9. Модельді оқыту

`model.fit` арқылы оқыт. `epochs=10`, `validation_data=test_ds`.
Нәтижені `history` айнымалысына сақта.

```python
# TODO: history = model.fit(train_ds, validation_data=test_ds, epochs=10)
history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=10
)
```

## Тапсырма 10. Дәлдік графигі және болжам

1) `history.history` ішіндегі `accuracy` мен `val_accuracy` графигін сыз.
2) Тест жиынынан бір батч ал, модельмен болжа және бірінші суретті
   нақты әрі болжанған класс атауымен көрсет.
Кеңес: `model.predict`, `np.argmax`.

```python
# 1-бөлім: оқыту графигі
# TODO: plt.plot(history.history['accuracy']) және val_accuracy
# TODO: plt.legend, plt.xlabel('epoch'), plt.show()
plt.plot(history.history['accuracy'], label='Оқу дәлдігі')
plt.plot(history.history['val_accuracy'], label='Тест дәлдігі')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.show()
# 2-бөлім: болжам
for images, labels in test_ds.take(1):
    preds = model.predict(images)
    pred_label = np.argmax(preds[0])

    plt.imshow(images[0])
    plt.title(
        "Нақты: " + class_names[int(labels[0])] +
        "\nБолжанған: " + class_names[pred_label]
    )
    plt.axis('off')
    plt.show()
```
